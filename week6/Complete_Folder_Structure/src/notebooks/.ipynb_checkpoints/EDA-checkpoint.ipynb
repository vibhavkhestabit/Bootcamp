```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Configuration for clearer plots
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Load the RAW dataset
# We use '../' to go up out of the notebooks folder
df = pd.read_csv('../data/raw/dataset.csv')

print(f"Dataset Shape: {df.shape}")
df.head()
```

```python
# 1. Calculate percentage of missing values
missing = df.isnull().sum()
missing_percent = (missing / len(df)) * 100
print("Missing Values (%):\n", missing_percent[missing_percent > 0])

# 2. Visual Heatmap 
plt.figure(figsize=(10, 6))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Missing Data Heatmap (Yellow = Missing)')
plt.show()
```

```python
# Countplot for Target Variable
sns.countplot(x='Survived', data=df, palette='pastel')
plt.title('Class Distribution: Survived (1) vs Died (0)')
plt.show()

# Exact numbers
print(df['Survived'].value_counts(normalize=True))
```

```python
# 1. Age Distribution (Histogram)
plt.figure(figsize=(10, 5))
sns.histplot(df['Age'].dropna(), kde=True, bins=30, color='blue')
plt.title('Age Distribution')
plt.show()

# 2. Fare Outliers (Boxplot) 

[Image of Boxplot for Outliers]

plt.figure(figsize=(10, 5))
sns.boxplot(x=df['Fare'], color='orange')
plt.title('Fare Boxplot (Check for Outliers)')
plt.show()
```

```python
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=[np.number])

# Heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Feature Correlation Matrix')
plt.show()
```
